```python
import PyPDF2
import re
import pandas as pd
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_extraction.text import TfidfVectorizer

import plotly.express as px
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
pdf_path = "pride-and-prejudice-jane-austen.pdf"

text_data = ""
page_count = 0

with open(pdf_path, "rb") as file:
    reader = PyPDF2.PdfReader(file)
    page_count = len(reader.pages)

    for page in reader.pages:
        text_data += page.extract_text()
```

```python
print("Total Pages in PDF:", page_count)
```

```python
print("Sample Extracted Text:\n")
print(text_data[:1000])
```

```python
# Convert to lowercase
text = text_data.lower()

# Remove numbers
text = re.sub(r'\d+', '', text)

# Remove special symbols
text = re.sub(r'[^\w\s]', '', text)

# Remove extra spaces
text = re.sub(r'\s+', ' ', text).strip()

print("Text After Regex Cleaning:\n")
print(text[:500])
```

```python
tokens = word_tokenize(text)
print("Total Tokens:", len(tokens))
print(tokens[:20])
```

```python
stop_words = set(stopwords.words('english'))

filtered_words = [word for word in tokens if word not in stop_words]

stopword_count = len(tokens) - len(filtered_words)

print("Total Stopwords Removed:", stopword_count)
print("Valid Words Count:", len(filtered_words))
```

```python
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("Sample Stemmed Words:\n", stemmed_words[:20])
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("Sample Lemmatized Words:\n", lemmatized_words[:20])
```

```python
sample_words = list(set(filtered_words[:20]))
encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform([[word] for word in sample_words])
onehot_df = pd.DataFrame(encoded, columns=encoder.categories_[0])
print("One Hot Encoding Output:\n")
onehot_df
```

```python
clean_text = " ".join(filtered_words)
vectorizer = TfidfVectorizer(max_features=50)
tfidf_matrix = vectorizer.fit_transform([clean_text])
feature_names = vectorizer.get_feature_names_out()
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names)
print("TF-IDF Feature Names:\n", feature_names)
tfidf_df
```

```python
tfidf_scores = tfidf_df.iloc[0]

plot_df = pd.DataFrame({
    "Word": tfidf_scores.index,
    "TF-IDF Score": tfidf_scores.values
})

fig = px.scatter(
    plot_df,
    x="Word",
    y="TF-IDF Score",
    title="TF-IDF Scatter Plot",
    labels={"Word": "Words", "TF-IDF Score": "Score"}
)

fig.show()
```
